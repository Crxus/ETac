```python
def rotate_mesh_euler(mesh, roll=0, pitch=0, yaw=0, degrees=True, order='xyz'):
    """
    使用欧拉角旋转网格
    
    参数:
        roll: 绕X轴旋转角度
        pitch: 绕Y轴旋转角度  
        yaw: 绕Z轴旋转角度
        degrees: 是否为角度制
        order: 旋转顺序，如 'xyz', 'zyx' 等
    """
    if degrees:
        roll, pitch, yaw = np.radians([roll, pitch, yaw])
    
    # 创建旋转矩阵
    R = Rotation.from_euler(order, [roll, pitch, yaw])
    
    # 创建4x4齐次变换矩阵
    transform = np.eye(4)
    transform[:3, :3] = R.as_matrix()
    
    # 应用变换
    mesh.apply_transform(transform)
    return mesh
```
